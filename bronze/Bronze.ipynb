```python
from pyspark.sql import SparkSession,functions as F,types as T

```

```python
spark=SparkSession.builder.getOrCreate()
```

```python
tables_list=['/Volumes/operation/bronze/raw_sources/CUST_AZ12.csv','/Volumes/operation/bronze/raw_sources/LOC_A101.csv','/Volumes/operation/bronze/raw_sources/PX_CAT_G1V2.csv','/Volumes/operation/bronze/raw_sources/cust_info.csv','/Volumes/operation/bronze/raw_sources/prd_info.csv','/Volumes/operation/bronze/raw_sources/sales_details.csv']
for table in tables_list:
  df=spark.read.csv(table,header=True)
  if table.split("/")[-1].split(".")[0] in ('CUST_AZ12','LOC_A101','PX_CAT_G1V2'):
      df.write.format('delta').mode('overwrite').saveAsTable(f'operation.bronze.ERP_{table.split("/")[-1].split(".")[0]}')
  else:
      df.write.format('delta').mode('overwrite').saveAsTable(f'operation.bronze.CRM_{table.split("/")[-1].split(".")[0]}')

```
